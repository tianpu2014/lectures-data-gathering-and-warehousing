## Key Concept 3: Data Stewardship Starts Before Analysis

A data workflow is not only a technical pipeline. It is also a stewardship decision.

Before using a dataset, ask:

- **Source:** Who published or collected it?
- **Purpose:** Why was the data collected?
- **Access:** Is it public, private, credentialed, scraped, or provided by an instructor?
- **Format:** CSV, JSON, HTML table, API response, database table, or another format?
- **Update pattern:** Is it static, periodically updated, or live?
- **Risks:** Does it include sensitive data, proxies for sensitive attributes, unclear terms, or misleading categories?
- **Verification:** What checks would prove the workflow is working?

These questions are part of technical work, not an afterthought.


## Tiny Demonstration: Build A Source-Idea Table

In Week 1, the table is intentionally tiny. The goal is to practice the complete evidence loop:

1. Create data.
2. Save data.
3. Reload data.
4. Verify data.
5. Explain the limitation.

This is the same pattern students will later use for public datasets, APIs, SQLite tables, and final projects.


```python
# Create a small source-idea table.
# Students can edit or add rows later in the notebook.

source_ideas = [
    {
        "topic": "Public service requests",
        "possible_source": "NYC Open Data 311 sample",
        "access_method": "Row-limited CSV or JSON API",
        "possible_question": "Which complaint types appear most often in a small recent sample?",
        "risk_or_limit": "A small sample cannot represent all neighborhoods or time periods."
    },
    {
        "topic": "Labor market trends",
        "possible_source": "BLS public API sample",
        "access_method": "No-key public API for one time series",
        "possible_question": "How does one unemployment series change over a short period?",
        "risk_or_limit": "One series does not explain causes of unemployment change."
    },
    {
        "topic": "Public health indicators",
        "possible_source": "CDC Open Data aggregated indicators",
        "access_method": "Small CSV or Socrata JSON query",
        "possible_question": "Which states have missing values for a selected indicator/year?",
        "risk_or_limit": "Aggregated indicators should not be interpreted as individual health records."
    },
    {
        "topic": "Government spending",
        "possible_source": "NJ Open Data purchasing sample",
        "access_method": "Small CSV/JSON extract",
        "possible_question": "Which agencies or vendors appear most often in a small sample?",
        "risk_or_limit": "Spending records need field definitions and context before interpretation."
    },
    {
        "topic": "Teaching dataset quality",
        "possible_source": "UCI Wine Quality dataset",
        "access_method": "Direct CSV download or instructor sample",
        "possible_question": "Which fields contain unusual ranges or require documentation?",
        "risk_or_limit": "Classic teaching datasets may not reflect current operational data collection."
    },
]

ideas_df = pd.DataFrame(source_ideas)
ideas_df
```

## Verification Cell: Did The Table Meet The Minimum Standard?

Verification is the habit that turns a notebook from a loose demonstration into evidence.

For this tiny table, minimum evidence means:

- At least five rows.
- Required columns are present.
- No missing source names.
- Every idea has a risk or limitation.

Later assignments will use the same pattern with real files, APIs, joins, and warehouse tables.


```python
required_columns = [
    "topic",
    "possible_source",
    "access_method",
    "possible_question",
    "risk_or_limit",
]

print("Rows:", len(ideas_df))
print("Columns:", list(ideas_df.columns))
print("Required columns present:", set(required_columns).issubset(ideas_df.columns))
print("Missing values by column:")
print(ideas_df[required_columns].isna().sum())

assert len(ideas_df) >= 5, "Expected at least 5 source ideas."
assert set(required_columns).issubset(ideas_df.columns), "Missing required columns."
assert ideas_df["possible_source"].notna().all(), "Every row needs a possible source."
assert ideas_df["risk_or_limit"].str.len().gt(0).all(), "Every row needs a risk or limitation."

print("Verification passed: the starter source table is complete enough for Week 1.")
```

## Save, Reload, And Verify Again

Saving and reloading seems simple, but it is a core reproducibility skill.

If a notebook creates an output but cannot reload it, the workflow is fragile. If a notebook reloads data but never checks row counts or columns, the instructor cannot tell whether the output is trustworthy.


```python
output_path = Path("week1_source_ideas.csv")
ideas_df.to_csv(output_path, index=False)

reloaded_df = pd.read_csv(output_path)

print("Saved file:", output_path)
print("Reloaded shape:", reloaded_df.shape)
print("Reloaded columns:", list(reloaded_df.columns))

assert len(reloaded_df) == len(ideas_df), "Reloaded row count does not match original."
assert list(reloaded_df.columns) == list(ideas_df.columns), "Reloaded columns do not match original."

reloaded_df.head()
```

## Mini-Concept: Functional And Nonfunctional Requirements

From the data warehousing lecture on reliable, scalable, and maintainable applications:

- **Functional requirements** describe what the system should do.
- **Nonfunctional requirements** describe qualities such as reliability, privacy, security, compliance, scalability, compatibility, and maintainability.

In this course, even a small notebook has both.

Example:

- Functional: load a CSV and summarize complaint types.
- Nonfunctional: run in Colab free tier, avoid private credentials, finish in under 10 minutes, show row-count checks, document source and limitation.

The nonfunctional requirements are part of the grade because they determine whether the data work can be trusted and reproduced.


```python
# Classify requirements for one tiny data workflow.

requirements = pd.DataFrame([
    {"requirement": "Load a small source-idea CSV", "type": "functional"},
    {"requirement": "Print row count after loading", "type": "nonfunctional / reliability"},
    {"requirement": "Avoid private credentials", "type": "nonfunctional / security"},
    {"requirement": "Keep runtime under 10 minutes", "type": "nonfunctional / feasibility"},
    {"requirement": "Document source and limitation", "type": "nonfunctional / stewardship"},
])

requirements
```
